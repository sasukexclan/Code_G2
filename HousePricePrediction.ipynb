```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="whitegrid")
plt.figure(figsize=(14, 6))
```

```python
# --- GRAPH 1: Distribution of Prices ---
# This shows the range of prices and where most houses are priced.
plt.subplot(1, 2, 1)
sns.histplot(df['price'], kde=True, color='teal')
plt.title('Distribution of House Prices')
plt.xlabel('Price')
plt.ylabel('Number of Houses')

# --- GRAPH 2: Price vs. Area ---
# This shows the correlation between the house size and its price.
plt.subplot(1, 2, 2)
sns.scatterplot(x='area', y='price', data=df, hue='airconditioning', alpha=0.7)
plt.title('How Price Varies with Area')
plt.xlabel('Area (sq ft)')
plt.ylabel('Price')

# Adjust layout and show the result
plt.tight_layout()
plt.show()
```
